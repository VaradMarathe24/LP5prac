```python
# ============================================================
# IMDB SENTIMENT ANALYSIS USING DEEP NEURAL NETWORK
# Practical Exam Ready Version with Better Visualizations
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

print("TensorFlow Version :", tf.__version__)

# ============================================================
# 2. LOAD IMDB DATASET
# ============================================================

vocab_size = 10000

(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(
    num_words=vocab_size
)

print("\nDataset Loaded Successfully")
print("Training Samples :", len(x_train))
print("Testing Samples  :", len(x_test))

# ============================================================
# 3. WORD INDEX DECODING
# ============================================================

word_index = keras.datasets.imdb.get_word_index()

reverse_word_index = {
    value + 3: key for key, value in word_index.items()
}

reverse_word_index[0] = "<PAD>"
reverse_word_index[1] = "<START>"
reverse_word_index[2] = "<UNK>"
reverse_word_index[3] = "<UNUSED>"

def decode_review(text):
    return " ".join(
        [reverse_word_index.get(i, "?") for i in text]
    )

# ============================================================
# 4. DISPLAY SAMPLE REVIEW
# ============================================================

print("\n================================================")
print("SAMPLE REVIEW")
print("================================================\n")

print(decode_review(x_train[0]))

print("\nActual Sentiment :",
      "Positive 😊" if y_train[0] == 1 else "Negative 😞")

# ============================================================
# 5. PAD SEQUENCES
# ============================================================

max_len = 256

x_train = keras.preprocessing.sequence.pad_sequences(
    x_train,
    value=0,
    padding='post',
    maxlen=max_len
)

x_test = keras.preprocessing.sequence.pad_sequences(
    x_test,
    value=0,
    padding='post',
    maxlen=max_len
)

print("\nSequences Padded Successfully")

# ============================================================
# 6. BUILD DNN MODEL
# ============================================================

model = keras.Sequential([

    layers.Input(shape=(max_len,)),

    # Word Embedding Layer
    layers.Embedding(vocab_size, 32),

    # Pooling Layer
    layers.GlobalAveragePooling1D(),

    # Deep Hidden Layers
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(32, activation='relu'),

    # Output Layer
    layers.Dense(1, activation='sigmoid')
])

# ============================================================
# 7. COMPILE MODEL
# ============================================================

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("\nMODEL SUMMARY")
print("================================================")

model.summary()

# ============================================================
# 8. VALIDATION SPLIT
# ============================================================

x_val = x_train[:10000]
partial_x_train = x_train[10000:]

y_val = y_train[:10000]
partial_y_train = y_train[10000:]

# ============================================================
# 9. TRAIN MODEL
# ============================================================

print("\n================================================")
print("TRAINING MODEL")
print("================================================")

history = model.fit(

    partial_x_train,
    partial_y_train,

    epochs=15,
    batch_size=512,

    validation_data=(x_val, y_val),

    verbose=1
)

# ============================================================
# 10. EVALUATE MODEL
# ============================================================

print("\n================================================")
print("MODEL EVALUATION")
print("================================================")

results = model.evaluate(x_test, y_test)

print("\nTest Loss     :", results[0])
print("Test Accuracy :", round(results[1] * 100, 2), "%")

# ============================================================
# 11. BEAUTIFUL VISUALIZATIONS
# ============================================================

history_dict = history.history

acc = history_dict['accuracy']
val_acc = history_dict['val_accuracy']

loss = history_dict['loss']
val_loss = history_dict['val_loss']

epochs = range(1, len(acc) + 1)

plt.figure(figsize=(16,6))

# ------------------------------------------------------------
# ACCURACY GRAPH
# ------------------------------------------------------------

plt.subplot(1,2,1)

plt.plot(
    epochs,
    acc,
    marker='o',
    linewidth=3,
    label='Training Accuracy'
)

plt.plot(
    epochs,
    val_acc,
    marker='*',
    linewidth=3,
    label='Validation Accuracy'
)

plt.title("Training vs Validation Accuracy", fontsize=15)
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.grid(True)
plt.legend()

# ------------------------------------------------------------
# LOSS GRAPH
# ------------------------------------------------------------

plt.subplot(1,2,2)

plt.plot(
    epochs,
    loss,
    marker='o',
    linewidth=3,
    label='Training Loss'
)

plt.plot(
    epochs,
    val_loss,
    marker='*',
    linewidth=3,
    label='Validation Loss'
)

plt.title("Training vs Validation Loss", fontsize=15)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

# ============================================================
# 12. CONFUSION MATRIX
# ============================================================

print("\n================================================")
print("CONFUSION MATRIX")
print("================================================")

predictions = model.predict(x_test)

y_pred = (predictions > 0.5).astype("int32")

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7,6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Negative', 'Positive'],
    yticklabels=['Negative', 'Positive']
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\n================================================")
print("CLASSIFICATION REPORT")
print("================================================")

print(classification_report(
    y_test,
    y_pred,
    target_names=['Negative', 'Positive']
))

# ============================================================
# 14. PREDICT MULTIPLE SAMPLE REVIEWS
# ============================================================

print("\n================================================")
print("SAMPLE REVIEW PREDICTIONS")
print("================================================")

sample_indices = [1]

for idx in sample_indices:

    sample_review = x_test[idx].reshape(1, max_len)

    prediction = model.predict(sample_review, verbose=0)[0][0]

    print("\n------------------------------------------------")
    print("Review Number :", idx + 1)
    print("------------------------------------------------")

    # Decode review text
    review_text = decode_review(x_test[idx][:120])

    print("\nReview Text:\n")
    print(review_text[:1000], "...")

    print("\nActual Sentiment :",
          "Positive 😊" if y_test[idx] == 1 else "Negative 😞")

    print("Prediction Score :", round(float(prediction), 4))

    if prediction > 0.5:
        print("Predicted Sentiment : Positive 😊")
    else:
        print("Predicted Sentiment : Negative 😞")

# ============================================================
# 15. CUSTOM USER REVIEW PREDICTION
# ============================================================
# ============================================================
# CUSTOM USER REVIEW PREDICTION (FIXED VERSION)
# ============================================================

from tensorflow.keras.preprocessing.sequence import pad_sequences
import re

print("\n================================================")
print("CUSTOM REVIEW PREDICTION")
print("================================================")

custom_review = """
This movie was absolutely fantastic.
The acting was brilliant and the story was amazing.
I really enjoyed watching this film.
"""

# Convert to lowercase
custom_review = custom_review.lower()

# Remove punctuation
custom_review = re.sub(r"[^\w\s]", "", custom_review)

# Convert words to integer sequence
sequence = []

for word in custom_review.split():

    index = word_index.get(word, 2)   # 2 = <UNK>

    # Add 3 because IMDB reserves indices 0-3
    sequence.append(index + 3)

# Add START token
sequence = [1] + sequence

# Pad sequence
padded = pad_sequences(
    [sequence],
    value=0,
    padding='post',
    maxlen=max_len
)

# Prediction
prediction = model.predict(padded, verbose=0)[0][0]

print("\nCustom Review:\n")
print(custom_review)

print("\nPrediction Score :", round(float(prediction), 4))

if prediction > 0.5:
    print("\nFinal Prediction : Positive Review 😊")
else:
    print("\nFinal Prediction : Negative Review 😞")

print("=====================")
print("End of the program")
print("====================")
```
